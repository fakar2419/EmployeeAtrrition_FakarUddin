<h2>Task 4 — Model Building & Comparison</h2>

<h4>Split data into training and test sets (80/20)</h4>

```python
from sklearn.model_selection import train_test_split

X = data.drop('Attrition', axis=1)
y = data['Attrition']

X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.20,random_state=42,stratify=y)

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("X_train shape:", X_train.shape)
print("X_test shape :", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape :", y_test.shape)
```

<h4>Handle class imbalance using class_weight='balanced' parameter (a simpler approach than SMOTE — good to know both exist)<br>
Train the following 3 models:<br>
Logistic Regression (baseline — also the most explainable to HR teams)<br>
Random Forest Classifier<br>
Gradient Boosting Classifier
</h4>

```python
# Logistic Regression
from sklearn.linear_model import LogisticRegression

lr_model = LogisticRegression(class_weight='balanced',C=0.07,solver='liblinear',max_iter=5000,random_state=42)

lr_model.fit(X_train_scaled, y_train)
```

```python
# Random Forest Classifier
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(
    n_estimators=500,
    max_depth=8,
    min_samples_split=10,
    min_samples_leaf=5,
    max_features='sqrt',
    class_weight='balanced',
    random_state=42
)

rf_model.fit(X_train, y_train)
```

```python
# Gradient Boosting Classifier
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.utils.class_weight import compute_sample_weight

sample_weights = compute_sample_weight(
    class_weight='balanced',
    y=y_train
)

gb_model = GradientBoostingClassifier(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=3,
    subsample=0.8,
    random_state=42
)

gb_model.fit(
    X_train,
    y_train,
    sample_weight=sample_weights
)
```

<h4>Record all results in a comparison table inside your notebook</h4>

```python
# Logistic Regression predictions
y_pred_lr = lr_model.predict(X_test_scaled)

# Random Forest predictions
y_pred_rf = rf_model.predict(X_test)

# Gradient Boosting predictions
y_pred_gb = gb_model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

results = pd.DataFrame({
    'Model': ['Logistic Regression',
              'Random Forest',
              'Gradient Boosting'],
    
    'Accuracy': [
        accuracy_score(y_test, y_pred_lr),
        accuracy_score(y_test, y_pred_rf),
        accuracy_score(y_test, y_pred_gb)
    ],
    
    'Precision': [
        precision_score(y_test, y_pred_lr),
        precision_score(y_test, y_pred_rf),
        precision_score(y_test, y_pred_gb)
    ],
    
    'Recall': [
        recall_score(y_test, y_pred_lr),
        recall_score(y_test, y_pred_rf),
        recall_score(y_test, y_pred_gb)
    ],
    
    'F1-Score': [
        f1_score(y_test, y_pred_lr),
        f1_score(y_test, y_pred_rf),
        f1_score(y_test, y_pred_gb)
    ]
})

results = results.round(4)
print(results)
```

<p>For employee attrition prediction, Recall and F1-score are more important than Accuracy because missing an employee who is likely to leave is costly for the company.Therefore Logistic Regression is the best model for this problem because it achieves the highest recall (68.09%) and the highest F1-score (48.12%), making it the most effective model for identifying employees at risk of leaving. It is also highly interpretable, which is beneficial for HR teams.</p>

<h2>Task 5 — Model Evaluation</h2>

<h2>Task 5 — Model Evaluation</h2>

<h4>Evaluate all 3 models using:<br>
Precision, Recall, F1-Score<br>
ROC-AUC Score<br>
Confusion Matrix
</h4>

```python
# Logistic Regression probabilities
y_prob_lr = lr_model.predict_proba(X_test_scaled)[:, 1]

# Random Forest probabilities
y_prob_rf = rf_model.predict_proba(X_test)[:, 1]

# Gradient Boosting probabilities
y_prob_gb = gb_model.predict_proba(X_test)[:, 1]
```

```python
from sklearn.metrics import  roc_auc_score
results = pd.DataFrame({
    'Model': ['Logistic Regression',
              'Random Forest',
              'Gradient Boosting'],
    
    'Precision': [
        precision_score(y_test, y_pred_lr),
        precision_score(y_test, y_pred_rf),
        precision_score(y_test, y_pred_gb)
    ],

    'Recall': [
        recall_score(y_test, y_pred_lr),
        recall_score(y_test, y_pred_rf),
        recall_score(y_test, y_pred_gb)
    ],

    'F1-Score': [
        f1_score(y_test, y_pred_lr),
        f1_score(y_test, y_pred_rf),
        f1_score(y_test, y_pred_gb)
    ],

    'ROC-AUC': [
        roc_auc_score(y_test, y_prob_lr),
        roc_auc_score(y_test, y_prob_rf),
        roc_auc_score(y_test, y_prob_gb)
    ]
})

results = results.round(4)
print(results)
```

```python
#Confusion Matrix

from sklearn.metrics import ConfusionMatrixDisplay
import matplotlib.pyplot as plt

models = {
    'Logistic Regression': (lr_model, X_test_scaled, y_pred_lr),
    'Random Forest': (rf_model, X_test, y_pred_rf),
    'Gradient Boosting': (gb_model, X_test, y_pred_gb)
}

for name, (model, X, y_pred) in models.items():
    ConfusionMatrixDisplay.from_predictions(y_test, y_pred)
    plt.title(name)
    plt.show()
```

<h4>Identify and clearly state which model performed best and why</h4>

<p>Best Model: Logistic Regression<br>
Logistic Regression was selected as the best-performing model. Although its accuracy (76.53%) is lower than that of Random Forest (82.65%) and Gradient Boosting (81.29), it achieved the highest Recall (68.09%) and the highest F1-Score (48.12%). Since the primary objective of employee attrition prediction is to identify employees who are likely to leave, Recall is more important than Accuracy. Logistic Regression correctly identifies a much larger proportion of employees who may leave the company, making it the most effective and practical model for HR decision-making. Additionally, Logistic Regression is highly interpretable, allowing HR teams to understand the factors influencing employee attrition.</p>

<h4>For the best model: extract Feature Importance — which factors matter most in predicting attrition?
</h4>

```python
#For Logistic Regression, we use the coefficients to determine feature importance.
feature_importance = pd.DataFrame({
    'Feature': X_train.columns,
    'Coefficient': lr_model.coef_[0]
})

feature_importance['Absolute_Coefficient'] = abs(
    feature_importance['Coefficient']
)

feature_importance = feature_importance.sort_values(
    by='Absolute_Coefficient',
    ascending=False
)

feature_importance
```

<h4>Rank the top 10 most important features driving employee exit</h4>

```python
top_10_features = (feature_importance.sort_values(by='Absolute_Coefficient', ascending=False).head(10))
top_10_features[['Feature', 'Coefficient']].reset_index(drop=True)
```

<h4>Chart 5 (Bonus): ROC Curve comparing all 3 models on one graph
</h4>

```python
from sklearn.metrics import roc_curve
# ROC curve coordinates
fpr_lr, tpr_lr, _ = roc_curve(y_test, y_prob_lr)
fpr_rf, tpr_rf, _ = roc_curve(y_test, y_prob_rf)
fpr_gb, tpr_gb, _ = roc_curve(y_test, y_prob_gb)

# ROC-AUC scores
auc_lr = roc_auc_score(y_test, y_prob_lr)
auc_rf = roc_auc_score(y_test, y_prob_rf)
auc_gb = roc_auc_score(y_test, y_prob_gb)

# Plot
plt.figure(figsize=(8, 6))

plt.plot(fpr_lr, tpr_lr,label=f'Logistic Regression (AUC = {auc_lr:.3f})')

plt.plot(fpr_rf, tpr_rf,label=f'Random Forest (AUC = {auc_rf:.3f})')

plt.plot(fpr_gb, tpr_gb,label=f'Gradient Boosting (AUC = {auc_gb:.3f})')

# Diagonal reference line
plt.plot([0, 1], [0, 1], 'k--', label='Random Guess')

plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve Comparison of All Models')
plt.legend(loc='lower right')

plt.savefig('charts/roc_curve_comparison.png',dpi=300,bbox_inches='tight')

plt.show()
```
